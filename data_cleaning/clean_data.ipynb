# Importação das bibliotecas

```python
import json
from pathlib import Path

import pandas as pd
```

# Configuração dos caminhos

```python
RAW_DIR = Path("../collector/data/raw")
PROCESSED_DIR = Path("../collector/processed")

FILES = [
    "spirit.json",
    "mouz.json",
    "falcons.json",
    "fut.json",
    "vitality.json",
    "furia.json"
]
```

# Função para carregar JSON

```python
def load_json(filepath):
    with open(filepath, "r", encoding="utf-8") as file:
        return json.load(file)
```

# Carregando os dados

```python
all_matches = []

for filename in FILES:

    filepath = f"{RAW_DIR}/{filename}"

    matches = load_json(filepath)

    for match in matches:
        match["_source"] = filename
        all_matches.append(match)

matches_df = pd.json_normalize(all_matches)

matches_df = matches_df.rename(
    columns={
        "id": "match_id",

        "team1.id": "team1_id",
        "team1.name": "team1_name",
        "team1.score": "team1_score",
        "team1.rank": "team1_rank",

        "team2.id": "team2_id",
        "team2.name": "team2_name",
        "team2.score": "team2_score",
        "team2.rank": "team2_rank",

        "winner.id": "winner_id",
        "winner.name": "winner_name"
    }
)
```

# Tratando a coluna de data

```python
matches_df["date"] = pd.to_datetime(
    matches_df["date"],
    errors="coerce"
)

matches_df["date"].dtype
```

# Tratando colunas numéricas

```python
numeric_columns = [
    "match_id",
    "team1_id",
    "team2_id",
    "team1_score",
    "team2_score",
    "team1_rank",
    "team2_rank",
    "winner_id",
    "best_of"
]

for column in numeric_columns:

    matches_df[column] = pd.to_numeric(
        matches_df[column],
        errors="coerce"
    )

matches_df = matches_df.drop_duplicates(
    subset="match_id"
).reset_index(drop=True)
```

# Verificando IDs duplicados

```python
print(
    f"Total de partidas: {len(matches_df)}"
)

print(
    f"IDs únicos: {matches_df['match_id'].nunique()}"
)

print(
    f"IDs duplicados: {matches_df['match_id'].duplicated().sum()}"
)

matches_df = matches_df.drop_duplicates(
    subset="match_id"
)
```

# Criando resultado das equipes

```python
def get_result(row, team_id):

    if row["winner_id"] == team_id:
        return "W"

    return "L"
# Resultado do Team 1
matches_df["team1_result"] = matches_df.apply(
    lambda row: get_result(
        row,
        row["team1_id"]
    ),
    axis=1
)
# Resultado do Team 2
matches_df["team2_result"] = matches_df.apply(
    lambda row: get_result(
        row,
        row["team2_id"]
    ),
    axis=1
)

# Verificando:

matches_df[
    [
        "team1_name",
        "team1_result",
        "team2_name",
        "team2_result"
    ]
].head()
```

# Criando diferença de rounds

```python
matches_df["score_difference"] = (
    matches_df["team1_score"]
    - matches_df["team2_score"]
).abs()
```

# Criando o DataFrame de mapas

```python
maps_data = []

for _, match in matches_df.iterrows():

    match_id = match["match_id"]

    for map_data in match["maps"]:

        maps_data.append({
            "match_id": match_id,
            "map_id": map_data["id"],
            "map_name": map_data["name"],
            "team1_map_score": map_data["team1_score"],
            "team2_map_score": map_data["team2_score"]
        })
```

# Criando diferença de rounds por mapa

```python
maps_df["round_difference"] = (
    maps_df["team1_map_score"]
    - maps_df["team2_map_score"]
).abs()
```
