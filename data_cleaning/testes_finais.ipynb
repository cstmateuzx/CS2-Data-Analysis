# Testes Gerais

```python
print("Matches:", matches_df.shape)
print("Maps:", maps_df.shape)
```

```python
matches_df["best_of"].value_counts().sort_index()
```

```python
maps_count = (
    maps_df
    .groupby("match_id")
    .size()
    .rename("maps_count")
    .reset_index()
)
```

```python
validation_df = matches_df[
    ["match_id", "best_of"]
].merge(
    maps_count,
    on="match_id",
    how="left"
)
```

```python
pd.crosstab(
    validation_df["best_of"],
    validation_df["maps_count"]
)
```
